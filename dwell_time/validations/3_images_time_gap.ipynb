```python
import sys, os; sys.path.append(os.path.join(os.path.dirname("2_images_time_gape.ipynb"), '..'))
from datetime import datetime, timedelta
from collections import defaultdict
import json
```

```python
ORIGIN_PKLOT = "/home/tteuz/Desktop/mestrado/meta_pklot/datasets/organized/PKLot"
ORIGIN_CNRPARK = "/home/tteuz/Desktop/mestrado/meta_pklot/datasets/organized/CNRParkEXT"
ORIGIN_PLDS = "/home/tteuz/Desktop/mestrado/meta_pklot/datasets/organized/PLds"
```

```python
jsons = [file for file in os.listdir(ORIGIN_PKLOT) if ".json" in file]

for json_file in jsons:
    with open(os.path.join(ORIGIN_PKLOT, json_file), "r") as file:
        data = json.load(file)
    
    timestamps = sorted([datetime(*image["date"], *image["time"]) for image in data["images"]])
    
    timestamps_by_day = defaultdict(list)
    for timestamp in timestamps:
        timestamps_by_day[timestamp.date()].append(timestamp)

    threshold = timedelta(minutes=90)
    gaps_by_day = defaultdict(list)

    for day, times in timestamps_by_day.items():
        prev_time = (times[0], 0)

        for index, time in enumerate(times[1:]):
            if (time - prev_time[0]) >= threshold:
                gaps_by_day[day].append((prev_time[1], index + 1, (time - prev_time[0])))
            
            prev_time = (time, index + 1)
    
    print(f"Subset: {json_file.replace('_spots.json', '').upper()}\n")
    for day, gaps in gaps_by_day.items():
        print(f"Day: {day} - Gaps: {' - '.join(f'[{gap[0]}, {gap[1]} - {gap[2]}]' for gap in gaps)}")
        print(f"Quantity of Gaps: {len(gaps)} - Images qty: {len(timestamps_by_day[day])}")
        print("")
```
