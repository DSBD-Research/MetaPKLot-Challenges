```python
import sys, os; sys.path.append(os.path.join(os.path.dirname("1_model_size.ipynb"), '..'))

from classification.models.mobilenet_v3_large import model
```

```python
MODEL_CONFIG = { "loss": "CrossEntropyLoss", "optimizer": "Adam", "device": "cuda:0" }
```

```python
def train_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

def all_parameters(model):
    return sum(p.numel() for p in model.parameters())
```

```python
test_model = model(pre_trained=True, config=MODEL_CONFIG)

print(f"Model size: {all_parameters(test_model.model)} - Train parameters: {train_parameters(test_model.model)}")
```
