```python
import sys, os; sys.path.append(os.path.join(os.path.dirname("1_images_avg_time_gap.ipynb"), '..'))
from collections import defaultdict
from datetime import datetime
import numpy as np
import json
```

```python
ORIGIN_PKLOT = "/home/tteuz/Desktop/mestrado/meta_pklot/datasets/organized/PKLot"
ORIGIN_CNRPARK = "/home/tteuz/Desktop/mestrado/meta_pklot/datasets/organized/CNRPark-EXT"
ORIGIN_PLDS = "/home/tteuz/Desktop/mestrado/meta_pklot/datasets/organized/PLds"
```

```python
jsons = [file for file in os.listdir(ORIGIN_PKLOT) if ".json" in file]

total_gaps = []
for json_file in jsons:
    with open(os.path.join(ORIGIN_PKLOT, json_file), "r") as file:
        data = json.load(file)
    
    timestamps = sorted([datetime(*image["date"], *image["time"]) for image in data["images"]])
    
    timestamps_by_day = defaultdict(list)
    for timestamp in timestamps:
        timestamps_by_day[timestamp.date()].append(timestamp)

    for day, times in timestamps_by_day.items():
        gaps = [(times[i] - times[i-1]).total_seconds() for i in range(1, len(times))]
        total_gaps.extend(gaps)

final_avg_gap, final_std_gap = np.average(total_gaps), np.std(total_gaps)
print(f"Final Average Gap: {final_avg_gap} seconds - Std Dev: {final_std_gap} seconds")
```
